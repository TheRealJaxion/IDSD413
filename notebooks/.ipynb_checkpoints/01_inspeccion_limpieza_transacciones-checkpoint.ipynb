# Fase 1: Inspeccion e Informacion del Dataset
## Importacion del Dataset

```python
# Cargamos librerias
import pandas as pd
import numpy as np
import matplotlib

# Rutas definidias:

ZIP_PATH = '../data/raw/data-card-detection.zip'
PROCESSED_PATH = '../data/processed'
FIGURES_PATH = '../figures'

df = pd.read_csv(ZIP_PATH, compression="zip")

#visualizamos

df.head()
```

Con esto, verificamos que el dataset ha sido correctamente cargado. 

## Informacion General del Dataset

Segun la descripcion de *Kaggle*, este dataset contiene **variables ya estandarizadas**, con $\bar{x} \approx 0$. El dataset ofrece a primera vista, las siguientes caracteristicas: 

| Tipo | Descripcion |
| :--- | :--- |
| *Time* (Fecha) | Tiempo (en segundos) transcurridos entre cada transaccion y la primera transaccion del dataset. |
| Variables *($V_1, V_2, ... , V_{28}$)* | Variables continuas **ortogonales** (sin correlacion alguna), producto del **Analisis de Componentes Principales (PCA)**, el cual es una tecnica de reduccion de dimensionalidad con el fin de proteger datos. | 
| *Amount* (Cantidad) | Cantidad de la transaccion. | 
| *Class* (Clase) | Contiene un valor binario ($Class = 0 ∨ Class = 1$), el cual indica cuando una transaccion no es fraudulenta ($Class = 0$) y cuando si lo es ($Class = 1$) |


Luego, hacemos una auditoria general al dataset para verificar: tipos correctos de datos, datos nulos y/o vacios y datos duplicados:

```python
# verificacion de datos nulos:
df.isnull().sum().sum()
```

```python
# toma un registro como original (keep='first') como referencia y las compara consecutivamente
duped_mask = df.duplicated(keep='first')

# calcula la cantidad de registros duplicados
total_duped = duped_mask.sum()

# Indexa etiquetas, tomando como argumentos la mascara y la columna clase
duped_fraud = df.loc[duped_mask, 'Class'].sum()
```

Luego, se procede a verificar la **distribucion** de los valores de la columna "Class", el cual indica posibles transacciones fraudulentas.

```python
# Conteo de cada dato (0 y 1)
class_count = df['Class'].value_counts()

# Distribucion porcentual de cada dato
dstr_class = df['Class'].value_counts(normalize=True) * 100
```

Una vez conocida esta informacion, procedemos a recopilarla a manera de un resumen.

## Resumen del Dataset

```python
print("=================================")
print("  RESUMEN GENERAL DEL DATASET")
print("=================================")
print("Ficha Tecnica de Extraccion: ")
print(f"Dimensiones del dataset: {df.shape[0]:,} filas; {df.shape[1]:,} columnas.")
print(f"Valores Blancos Totales: {df.isna().sum().sum()}")
print(f"Valores Duplicados: {total_duped}")
print(f"Duplicados Fraude: {duped_fraud}")
print("=================================")
print("Tipos de Datos: ")
print(f"{df.dtypes.value_counts()}")
print("=================================")
print("Distribucion de Clases (Class): ")
print(f" - Legitimas (Class = 0): {class_count[0]:,} correspondientes al {dstr_class[0]:.3f}% de los datos.")
print(f" - Fraudulentas (Class = 1): {class_count[1]:,} correspondientes al {dstr_class[1]:.3f}% de los datos.")
```

El dataset esta compuesto por 284,807 filas, y 31 columnas (30 variables del tipo `float64` y una variable del tipo entera `int64`), ninguna de ellas con valores nulos o faltantes (`NaN`). De igual manera, existen 1081 registros duplicados, de los cuales 19 son fraudulentas. Por ultimo, se observa que el $99,827\%$ de los datos son categorizadas como transacciones legitimas (`Class = 0`), mientras que el $0.173\%$ se categorizan como fraudulentas (`Class = 1`).

# Fase 2: Limpieza Estructural y Persistencia del Dataset Depurado
A partir de los datos recopilados en la inspeccion y auditoria inicial, se procede a realizar la limpieza y depuracion del *pipeline*.

## Eliminacion de Registros Duplicados
Los valores de la varianza $s^2$ se encuentra sesgado en el dataset actual, dado que esta trabajando con un tamaño muestral $n$ alterado artificialmente por sus registros duplicados. Para evitar distorsiones tanto en sus grados de libertad, como una subestimacion en los estadisticos, es necesario eliminar estos datos.

```python
# Documentamos la cantidad de filas iniciales
initial_rows = len(df)

# Eliminacion segura
df_clean = df.drop_duplicates(keep='first').reset_index(drop=True)

# Filas tras la eliminacion
final_rows = len(df_clean)
```

## Downcasting
Dado a la cantidad masiva de datos con la que se esta trabajando, se puede optimizar el uso de memoria haciendo *Downcasting* a las variables del dataset, sin perder precision dado a su integridad inicial

```python
# Tomamos columnas del tipo float64
cols_c = [col for col in df_clean.columns if col != 'Class']
df_clean[cols_c] = df_clean[cols_c].astype('float32')
df_clean['Class'] = df_clean['Class'].astype('int8')
```

```python
# Por ultimo, se verifica la informacion y se genera un reporte:
dupes_rem = df_clean.duplicated().sum()
clean_count = df_clean['Class'].value_counts()
perc_cleaned = df_clean['Class'].value_counts(normalize=True) * 100

print("==========================================================")
print("          REPORTE DE LIMPIEZA Y DEPURACIÓN")
print("==========================================================")
print(f"Registros iniciales:        {initial_rows:,}")
print(f"Registros eliminados:       {initial_rows  - final_rows:,}")
print(f"Registros limpios finales:  {initial_rows :,}")
print(f"Duplicados restantes:       {dupes_rem}")
print(f"Nueva huella en memoria:    {df_clean.memory_usage().sum() / 1024**2:.2f} MB")
print("----------------------------------------------------------")
print("Nueva distribución de Clases:")
print(f"  - Legítimas (Class = 0):  {clean_count [0]:,} ({perc_cleaned[0]:.3f}%)")
print(f"  - Fraudes   (Class = 1):  {clean_count [1]:,} ({perc_cleaned[1]:.3f}%)")
print("==========================================================")
```

Se observa que se han eliminado exactamente 1081 registros duplicados. A su vez, al hacer Downcasting en las variables se ha ahorrado 32.74MB de memoria. La distribucion en las variables que se categorizaban como legitimas y fraudulentas han cambiando, con apenas una diferencia del 0.06% de los datos.

## Persistencia
Por ultimo, se procede a guardar el archivo limpio en la ruta **"../data/processed"** (PROCESSED_PATH) para documentar correctamente los cambios realizados en el dataset. Al tratarse de un dataset muy extenso (200mil registros), es una buena practica el usar el paquete **PyArrow** (Apache Parquet) para comprimir, almacenar y leer el archivo de manera eficiente, estructurada y optimizada.

```python
# Guardamos los archivos en extension .parquet y .gz
parquet_out = f"{PROCESSED_PATH}/transactions_cleaned.parquet"
csv_out = f"{PROCESSED_PATH}/transactions_cleaned.csv.gz"

df_clean.to_parquet(parquet_out, index=False, engine="pyarrow")
df_clean.to_csv(csv_out, index=False, compression="gzip")
```
